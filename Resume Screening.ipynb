```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("resumes.csv")

df.head()
```

```python
df.info()

df.isnull().sum()
```

```python
ranked_candidates = df.sort_values(by="Score", ascending=False)

ranked_candidates
```

```python
top5 = ranked_candidates.head(5)

print(top5)
```

```python
role = "Data Scientist"

filtered = df[df["Job_Role"] == role]

filtered
```

```python
plt.figure(figsize=(8,5))

plt.bar(df["Resume_ID"], df["Score"])

plt.title("Candidate Ranking by Resume Score")
plt.xlabel("Resume ID")
plt.ylabel("Score")

plt.show()
```
